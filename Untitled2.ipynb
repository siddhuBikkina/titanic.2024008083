```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Titanic.csv')
```

```python
df.head()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
print(df.columns)
```

```python
df.info()
```

```python
numerical_columns = df.select_dtypes(include=np.number).columns

print("Numerical Columns:")
print(numerical_columns)
```

```python
numeric_df = df[numerical_columns]

numeric_df.head()
```

```python
print(df[numerical_columns].isnull().sum())
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
mean = numeric_df.mean()

print("Mean:")
print(mean)
```

```python
data_range = numeric_df.max() - numeric_df.min()

print("Range:")
print(data_range)
```

```python
variance = numeric_df.var()

print("Variance:")
print(variance)
```

```python
std_deviation = numeric_df.std()

print("Standard Deviation:")
print(std_deviation)
```

```python
Q1 = numeric_df.quantile(0.25)
Q2 = numeric_df.quantile(0.50)
Q3 = numeric_df.quantile(0.75)

print("Q1:")
print(Q1)

print("\nQ2 (Median):")
print(Q2)

print("\nQ3:")
print(Q3)
```

```python
IQR = Q3 - Q1

print("Interquartile Range:")
print(IQR)
```

```python
mad = numeric_df.apply(lambda x: np.mean(np.abs(x - x.mean())))

print("Mean Absolute Deviation:")
print(mad)
```

```python
cv = (numeric_df.std() / numeric_df.mean()) * 100

print("Coefficient of Variation (%):")
print(cv)
```

```python
dispersion = pd.DataFrame({
      'Mean': numeric_df.mean(),
          'Minimum': numeric_df.min(),
              'Maximum': numeric_df.max(),
                  'Range': numeric_df.max() - numeric_df.min(),
                      'Variance': numeric_df.var(),
                          'Standard Deviation': numeric_df.std(),
                              'Q1': numeric_df.quantile(0.25),
                                  'Median': numeric_df.median(),
                                      'Q3': numeric_df.quantile(0.75),
                                          'IQR': numeric_df.quantile(0.75) - numeric_df.quantile(0.25),
                                              'MAD': numeric_df.apply(lambda x: np.mean(np.abs(x - x.mean()))),
                                                  'CV (%)': (numeric_df.std() / numeric_df.mean()) * 100
})

dispersion
```

```python
five_number_summary = pd.DataFrame({
      'Minimum': numeric_df.min(),
          'Q1': numeric_df.quantile(0.25),
              'Median': numeric_df.median(),
                  'Q3': numeric_df.quantile(0.75),
                      'Maximum': numeric_df.max()
})
```

```python
plt.figure(figsize=(12, 6))

sns.boxplot(data=numeric_df)

plt.title("Box Plot of Titanic Numerical Features")
plt.xticks(rotation=45)
plt.show()
```

```python
numeric_df.hist(figsize=(15, 10), bins=20)

plt.suptitle("Distribution of Titanic Numerical Features")
plt.show()
```

```python
mean_age = df['Age'].mean()
std_age = df['Age'].std()

print("Mean Age:", mean_age)
print("Standard Deviation of Age:", std_age)

plt.figure(figsize=(10, 5))

plt.hist(df['Age'], bins=20)

plt.axvline(mean_age, linestyle='--', linewidth=2)
plt.axvline(mean_age + std_age, linestyle='--', linewidth=2)
plt.axvline(mean_age - std_age, linestyle='--', linewidth=2)

plt.title("Age Distribution with Mean and Standard Deviation")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
Q1 = numeric_df.quantile(0.25)
Q3 = numeric_df.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:")
print(lower_bound)

print("\nUpper Bound:")
print(upper_bound)
```

```python
outliers = ((numeric_df < lower_bound) | (numeric_df > upper_bound)).sum()

print("Number of Outliers:")
print(outliers)
```

```python
correlation = numeric_df.corr()

correlation
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(correlation, annot=True, cmap='coolwarm')

plt.title("Correlation Heatmap of Titanic Dataset")

plt.show()
```
