```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Automobile_data.csv")
```

```python
df.columns = df.columns.str.strip()
```

```python
df.replace("?", pd.NA, inplace=True)
```

```python
numeric_cols = [
    "price", "horsepower", "engine-size",
    "city-mpg", "highway-mpg", "curb-weight"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Remove rows with missing values in important columns
df.dropna(subset=numeric_cols, inplace=True)

sns.set_style("whitegrid")
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["price"], bins=15, edgecolor="black")
plt.title("Distribution of Automobile Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="body-style", y="price", errorbar=None)
plt.title("Average Price by Body Style")
plt.xlabel("Body Style")
plt.ylabel("Average Price")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="engine-size", y="price")
plt.title("Engine Size vs Price")
plt.xlabel("Engine Size")
plt.ylabel("Price")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="fuel-type", y="price")
plt.title("Price Distribution by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Price")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="body-style")
plt.title("Number of Cars by Body Style")
plt.xlabel("Body Style")
plt.ylabel("Number of Cars")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(range(len(df)), df["price"])
plt.title("Automobile Price Trend")
plt.xlabel("Record Number")
plt.ylabel("Price")
plt.show()
```

```python
fuel_counts = df["fuel-type"].value_counts()
plt.figure(figsize=(7, 7))
plt.pie(
    fuel_counts,
    labels=fuel_counts.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Distribution of Fuel Types")
plt.show()
```

```python
plt.figure(figsize=(9, 6))
corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Automobile Features")
plt.show()
```

```python
sns.pairplot(
    df[["price", "horsepower", "engine-size", "city-mpg"]],
    diag_kind="hist"
)
plt.show()
```
