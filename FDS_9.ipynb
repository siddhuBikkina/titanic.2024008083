```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Iris dataset
df = pd.read_csv('/content/iris.csv')

# Display first five records
print(df.head())

# Display dataset information
print(df.info())

# Check for missing values
print(df.isnull().sum())
```

```python
df.groupby('species')[[
    'sepal_length', 'sepal_width',
    'petal_length', 'petal_width'
]].mean().plot(kind='bar', figsize=(10, 5))

plt.title('Average Measurements of Iris Species')
plt.xlabel('Species')
plt.ylabel('Average Measurement (cm)')
plt.xticks(rotation=0)
plt.legend(title='Features')
plt.tight_layout()
plt.show()
```

```python
plt.hist(df['sepal_length'], bins=10, edgecolor='black')

plt.title('Distribution of Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x='sepal_length',
    y='petal_length',
    hue='species'
)

plt.title('Sepal Length vs Petal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')
plt.show()
```

```python
sns.pairplot(df, hue='species')
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df[['sepal_length', 'sepal_width',
             'petal_length', 'petal_width']]
)

plt.title('Box Plot of Iris Measurements')
plt.xlabel('Features')
plt.ylabel('Measurement (cm)')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

corr = df.select_dtypes(
    include='number'
).corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap of Iris Features')
plt.show()
```

```python
sns.countplot(data=df, x='species')

plt.title('Count of Iris Species')
plt.xlabel('Species')
plt.ylabel('Number of Flowers')
plt.show()
```

```python
df['species'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(6, 6)
)

plt.title('Percentage of Iris Species')
plt.ylabel('')
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    df.index,
    df['sepal_length'],
    label='Sepal Length'
)

plt.plot(
    df.index,
    df['petal_length'],
    label='Petal Length'
)

plt.title('Sepal Length and Petal Length')
plt.xlabel('Flower Index')
plt.ylabel('Length (cm)')
plt.legend()
plt.show()
```

```python
sns.kdeplot(
    data=df,
    x='petal_width',
    hue='species',
    fill=True
)

plt.title('Density Distribution of Petal Width')
plt.xlabel('Petal Width (cm)')
plt.show()
```
